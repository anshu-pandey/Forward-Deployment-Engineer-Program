3. CRUD Operations
CRUD stands for:

- C - Create - Add new books to the list.
- R - Read - Retrieve and display books from the list.
- U - Update - Modify the information of existing books.
- D - Delete - Remove books from the list.

These four operations form the foundation of most applications that manage data.

4. Understanding Request and Response
A web application generally has two main components:

- Frontend (Web Page): Where users interact with the application.
- Backend (FastAPI Server): Processes requests and handles business logic.



The communication process:

- The user interacts with the frontend.
- The frontend sends an HTTP request to the FastAPI server.
- FastAPI processes the request.
- FastAPI sends an HTTP response.
- The frontend displays the returned information.

Example:

- Suppose the frontend requests Book 2 and Book 4.
- Request: Give me Book 2 and Book 4.
- FastAPI: Finds the requested books.
- Response: Returns Book 2 and Book 4 to the frontend.


5. What Are HTTP Request Methods?
HTTP request methods (also called HTTP verbs) tell the server what action the client wants to perform.

For example:

- Retrieve information.
- Create new information.
- Update existing information.
- Delete information.
- FastAPI uses these HTTP methods to perform CRUD operations.

6. Mapping CRUD Operations to HTTP Methods
This is one of the most important concepts in this lecture.



Explanation with examples
1. POST -> Create -> Used to add a new book. -> http POST /books
2. GET -> Read -> Used to retrieve books. -> http GET /books
3. PUT ->  Update -> Used to update an existing book. -> http PUT /books/2
4. DELETE -> Delete -> Used to delete a book. -> http DELETE /books/2
Note: PUT is generally used to replace a resource. PATCH is commonly used for partial updates.



7. What Is Swagger UI?
FastAPI provides an automatic interactive API documentation interface called Swagger UI.

Swagger UI allows developers to:
- View all available API endpoints.
- See supported HTTP request methods.
- Test API endpoints directly from the browser.
- Send requests to the FastAPI application.
- View API responses.

```python
"""Explanation:

fastapi is the Python package.
FastAPI is the class used to create our application.
Importing it allows us to use FastAPI functionality in our project.
"""
from fastapi import Body,FastAPI
```

```python
"""

Explanation:

FastAPI() creates a new FastAPI application.
app is the variable that stores our application instance.
We use this app object to define API endpoints and configure our application.
This is the starting point of almost every FastAPI project.

"""
app = FastAPI()
```

```python
"""Create Your First GET API Endpoint 


Understanding the code
1. @app.get("/api-endpoint")

This is called a decorator.

It tells FastAPI that the function below it should handle GET requests.

/api-endpoint is the URL path of the endpoint.

2. async def first_api():

Defines an asynchronous Python function.

first_api is the function name.

async allows asynchronous execution.

3. return {"message": "Hello Eric"}

Returns a Python dictionary.

FastAPI automatically converts the dictionary into a JSON response.


"""

@app.get("/api-endpoint")
async def first_api():
    return {"message": "Hello Eric"}


"""What is async?
async stands for asynchronous.

It allows a function to perform asynchronous operations.

FastAPI supports both normal def and asynchronous async def endpoint functions.

Using async is not compulsory.

It is particularly useful when working with asynchronous operations, such as database or network calls.

Important: FastAPI does not automatically make every function asynchronous. You should use async def when your code uses asynchronous operations.

"""
```

What Is an API Endpoint?
An API endpoint is a specific URL path through which a client communicates with the backend application.

For example:
http://127.0.0.1:8000/api-endpoint



When a client sends a GET request to this URL, FastAPI executes the associated Python function and returns its response.

**Path Parameters**

1. What are Path Parameters?
Path parameters are variables included directly in a URL that allow us to retrieve specific resources.

They are used when we want to identify or access particular information based on its location in the URL.

For example:
http://127.0.0.1:8000/books/book1

Here:

- /books is the static path.
- /book1 is the dynamic path parameter.

The user can change book1 to book2, book3, etc., to retrieve different books.

2.  Static Paths vs Dynamic Paths
A. Static Path : 
    A static path is fixed and does not change.

Example:



URL:
http://127.0.0.1:8000/books
Whenever a user visits /books, the same endpoint is called.


B. Dynamic Path : 
    A dynamic path allows users to pass different values through the URL.

Example:



Now we can pass different values:

URL  -> Response

/books/book1 -> book1

/books/book2 -> book2

/books/book3 -> book3

/books/book4 -> book4

We don't need to create a separate API endpoint for every book.
One dynamic endpoint can handle multiple values.


3. How Do Dynamic Path Parameters Work?



There are two important parts:

1. {dynamic_param}

Curly brackets indicate a dynamic path parameter.

FastAPI captures the value provided in that position of the URL.

2. dynamic_param

This is the function parameter.

FastAPI passes the captured URL value into this variable.

For example, when the user requests:
http://127.0.0.1:8000/books/book1

FastAPI extracts book1 and passes it into the Python function.

Response:

json
{
    "dynamic_param": "book1"
}


Important rule: The parameter name inside the curly brackets must match the corresponding Python function parameter name.



4. Order Matters in Path Parameters
This is a very important concept in FastAPI.

FastAPI checks matching routes in the order they are registered.

Suppose we have:


Now imagine the user visits:
http://127.0.0.1:8000/books/mybook

What happens?

- FastAPI checks the first endpoint.
- /books/{dynamic_param} matches /books/mybook.
- mybook gets captured as dynamic_param.
- The first function executes.
- The second endpoint never gets a chance to handle that request.



Correct approach: Static routes first
Place specific static routes before dynamic routes.



Now ,



Always define specific static routes before general dynamic routes when they could match the same URL.
Fast API works in chronological order
