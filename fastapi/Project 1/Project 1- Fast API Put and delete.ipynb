**PUT HTTP Request Method in FastAPI**

A PUT request is used to update an existing resource on the server.

In this example, we have a list of books. We want to update an existing book's information, such as its author or category, without creating a new book.



**Implementing PUT in FastAPI**

```python
@app.put("/books/update_book")
async def update_book(updated_book=Body()):
    for i in range(len(BOOKS)):
        if BOOKS[i].get('title').casefold() == updated_book.get('title').casefold():
            BOOKS[i] = updated_book
            break
```

**4. Important concepts to remember**
- POST: Creates a new resource.
- PUT: Replaces an existing resource with the supplied representation.
- PATCH: Updates only specific fields of an existing resource.
One important distinction: In this tutorial, the PUT implementation replaces the entire book object, not just the category. That's why the request body includes the title, author, and category.

A DELETE request can technically have a request body, but it is generally not recommended.
1. Can DELETE have a body?
HTTP does not explicitly prohibit a DELETE request body. However, its semantics are not clearly defined by the HTTP standard, and some servers, proxies, and clients may not support it consistently.

Yes, an HTTP DELETE request can technically have a body, though you should generally avoid sending one.

- What the HTTP Specification Says

• No defined semantics. According to the HTTP Semantics specification (RFC 9110) and MDN Web Docs, content inside a DELETE request has no generally defined meaning and cannot alter the target of the request.
• Discouraged for clients. The specification states that clients should not generate a body in a DELETE request unless the server explicitly indicates support for it.

- Practical Risks & Real-World Behavior

• Silent drops and rejections. Because many servers, proxies, firewalls, and caching layers expect DELETE requests to be bodyless, they may silently strip the payload or reject the request with a 400 Bad Request or 413 Payload Too Large error.

• Framework friction. While some tools like AWS API Gateway allow it, many web servers and frameworks (such as Nginx, Apache, Django, or Express) either ignore DELETE bodies by default or require custom middleware to parse them.

• Developer debate. On Stack Overflow, opinions are mixed: some developers use DELETE bodies for bulk deletions or concurrency tokens, while others warn against it due to infrastructure incompatibilities and request-smuggling risks
