Q1) What is FastAPI?
FastAPI is a modern Python web framework used to build APIs.

It is designed to make API development:

- Simple
- Fast
- Scalable
- Extensible

It helps developers build RESTful APIs with minimal code.

FastAPI handles many common API development tasks automatically, so developers don't need to reinvent the wheel.

Q2) Why is it called FastAPI?
The word Fast has two meanings:

A. Fast Performance
FastAPI is one of the fastest-performing Python web frameworks.
- It provides high performance.
- It includes built-in features such as:
    - Data validation
    - Data serialization
    - Automatic API documentation
    - Security support

B. Fast Development
Developers can build APIs with just a few lines of code.
- It reduces development time.
- It handles much of the work required to create RESTful APIs.
- APIs can be created quickly and prepared for deployment.

Q4) Where Does FastAPI Fit in Application Architecture?
FastAPI acts as the backend server application in a typical web application.

A typical application has two main parts:
    
    A. Web Page (Frontend)
        - This is what users interact with.
        - It displays information and collects user input.
        - It sends requests to the backend through RESTful APIs.

    B. Server (Backend - FastAPI)
        - FastAPI runs on the server side.
        - It handles the business logic of the application.
        - It processes requests received from the frontend.
        - It communicates with databases and other services when needed.
        - It sends appropriate responses back to the frontend.

How do they communicate?

    - The user interacts with the web page.
    - The frontend sends an API request to the FastAPI backend.
    - FastAPI processes the request and executes the required business logic.
    - FastAPI returns the response.
    - The frontend displays the received data to the user.

Example:

- Suppose a user logs into an application:
- Frontend sends the username and password to FastAPI.
- FastAPI validates the credentials.
- FastAPI checks the user information.
- FastAPI sends an appropriate response to the frontend.

Q5) FastAPI for Full-Stack Applications
FastAPI can be used in two ways:

- RESTful API applications: FastAPI handles backend logic and communicates with a separate frontend.

- Full-stack applications: FastAPI can also render frontend web pages along with handling backend logic.
